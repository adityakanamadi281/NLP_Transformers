```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
train_sentence = [ "I love playing football",
    "I love playing chess",
    "I enjoy reading books",
    "She loves cooking pasta",
    "He likes watching movies",
    "They are playing cricket",
    "We enjoy learning Python",
    "You love listening to music"]
```

```python
train_sentence
```

```python
tokenizer=Tokenizer(num_words=100)
```

```python
tokenizer
```

```python
tokenizer.fit_on_texts(train_sentence)
```

```python
word_index=tokenizer.word_index
```

```python
word_index
```

# Create Sequence

```python
sequences = tokenizer.texts_to_sequences(train_sentence)
```

```python
print(f"word index {word_index}")
print(f"sequence of words {sequences}")
```

```python
print(train_sentence[0])
print(sequences[0])
```
